```python
# Fichier: ExplorationDonnées.ipynb
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python
# Configurer l'affichage pour les notebooks
%matplotlib inline
sns.set(style="whitegrid", color_codes=True)
```


 1. Configuration des Chemins

```python

PROJECT_ROOT = Path("..")
DATA_RAW_PATH = PROJECT_ROOT / "data" / "raw"

DATA_FILE = "cardiaque.csv"
DATA_FILE_PATH = DATA_RAW_PATH / DATA_FILE

print(f"Chemin du projet racine : {PROJECT_ROOT.resolve()}")
print(f"Tentative de chargement du fichier : {DATA_FILE_PATH.resolve()}")
```

```python
# =============================================================================
# 2. Chargement des Données
# =============================================================================
try:
    df = pd.read_csv(DATA_FILE_PATH)
    print("Données chargées avec succès.")
except FileNotFoundError:
    print(f"ERREUR : Le fichier {DATA_FILE_PATH} n'a pas été trouvé.")
    print("Veuillez vérifier que le fichier est bien dans le dossier 'data/raw/'.")
```

```python
# =============================================================================
# 3. Inspection Initiale des Données
# =============================================================================
if 'df' in locals():
    print("\n--- 3. Inspection Initiale ---")
    
    # Afficher les 5 premières lignes
    print("\nAperçu des données (head) :")
    display(df.head())
    
    # Obtenir les informations sur les colonnes (types, non-null)
    print("\nInformations sur les colonnes (info) :")
    df.info()
    
    # Obtenir les statistiques descriptives pour les colonnes numériques
    print("\nStatistiques descriptives (describe) :")
    display(df.describe())
    
    # Vérifier les valeurs manquantes
    print("\nTotal des valeurs manquantes par colonne :")
    print(df.isnull().sum())
```

```python
"""
**age** : L'âge du patient en années. 
**sex** : Le sexe du patient (généralement codé comme 1 = homme, 0 = femme). 
**cp** (chest pain type) : Type de douleur thoracique (valeurs typiques incluent 1 = angine typique, 2 = angine atypique, 3 = douleur non angineuse, 4 = asymptomatique). 
**trestbps** (resting blood pressure) : La tension artérielle au repos en mm Hg (millimètres de mercure) lors de l'admission à l'hôpital. 
**chol** (serum cholesterol) : Le cholestérol sérique en mg/dl (milligrammes par décilitre). 
**fbs** (fasting blood sugar) : Le taux de sucre dans le sang à jeun > 120 mg/dl (1 = vrai ; 0 = faux). 
**restecg** (resting electrocardiographic results) : Résultats de l'électrocardiogramme au repos (valeurs telles que 0 = normal, 1 = ayant des anomalies de l'onde ST-T, 2 = montrant une hypertrophie ventriculaire gauche probable ou définitive selon les critères d'Estes). 
**thalach** (maximum heart rate achieved) : Fréquence cardiaque maximale atteinte. 
**exang** (exercise induced angina) : Angine déclenchée par l'exercice (1 = oui ; 0 = non). 
**oldpeak** (ST depression induced by exercise relative to rest) : Dépression du segment ST induite par l'exercice par rapport au repos. 
**slope** (the slope of the peak exercise ST segment) : La pente du segment ST d'effort (valeurs telles que 1 = pente ascendante, 2 = plat, 3 = pente descendante). 
**ca** (number of major vessels colored by flourosopy) : Nombre de vaisseaux principaux (0-3) colorés par fluoroscopie. 
**thal** (thalassemia) : Une anomalie de l'hémoglobine (3 = normal ; 6 = défaut fixe ; 7 = défaut réversible). 
**num** : Présence de maladie cardiaque (1 = présence, 0 = absence).
"""
```

```python
# =============================================================================
# 4. Nettoyage des Données
# =============================================================================

if 'df' in locals():
    print("\n--- 3.6 Nettoyage (Valeurs Manquantes et Colonnes Inutiles) ---")


    ###################""
    
    # 1. Supprimer la colonne 'Unnamed: 0' (index inutile)
    if 'Unnamed: 0' in df.columns:
        df = df.drop(columns=['Unnamed: 0'])
        print("Colonne 'Unnamed: 0' supprimée.")
    else:
        print("Colonne 'Unnamed: 0' non trouvée (déjà supprimée).")

    # 2. Gérer les valeurs manquantes (Suppression des lignes)
    lignes_avant = df.shape[0]
    print(f"Nombre de lignes avant suppression des NaN : {lignes_avant}")

    # On supprime les lignes où 'ca' OU 'thal' est manquant
    df = df.dropna(subset=['ca', 'thal'])

    # On supprime des doublons si présents
    df = df.drop_duplicates()

    lignes_apres = df.shape[0]
    print(f"Nombre de lignes après suppression des NaN : {lignes_apres}")
    print(f"-> {lignes_avant - lignes_apres} lignes ont été supprimées.")

    # 3. Vérification finale
    print("\nVérification des valeurs manquantes (après correction) :")
    print(df.isnull().sum().to_string())
else:
    print("ERREUR : 'df' n'a pas été chargé.")
```

```python
# =============================================================================
# 5. Analyse de la Variable Cible ('target')
# =============================================================================
# La variable cible est 'target' selon le dataset
if 'df' in locals():
    print("\n--- 5. Analyse de la Variable Cible ---")
    TARGET_COL = 'num'
    
    # Vérifier l'équilibre des classes
    print(f"\nDistribution de la variable cible '{TARGET_COL}' :")
    print(df[TARGET_COL].value_counts(normalize=True))
    
    # Visualisation de l'équilibre
    plt.figure(figsize=(6, 4))
    sns.countplot(x=TARGET_COL, data=df)
    plt.title("Distribution de la variable cible (0 = Sain, 1 = Malade)")
    plt.show()
```

```python
# =============================================================================
# 6. Identification des Types de Caractéristiques (Features)
# =============================================================================
# Basé sur le dictionnaire de données Kaggle, nous identifions
# manuellement les colonnes numériques et catégorielles.
# 'id' est retiré car ce n'est pas une feature.

if 'df' in locals():
    print("\n--- 6. Analyse des Caractéristiques ---")
    
    # Exclure l'ID et la cible pour l'analyse des features
    features_df = df.drop(columns=['id', TARGET_COL], errors='ignore')
    
    # Définition manuelle basée sur le dictionnaire de données
    numerical_features = ['age', 'trestbps', 'chol', 'thalach', 'oldpeak']
    
    # Toutes les autres features sont catégorielles (même si encodées en int)
    categorical_features = [
        col for col in features_df.columns 
        if col not in numerical_features
    ]
    
    print(f"Features numériques (continues) : {numerical_features}")
    print(f"Features catégorielles (discrètes) : {categorical_features}")
```

```python
# =============================================================================
# 7. Visualisation des Caractéristiques
# =============================================================================
if 'df' in locals():
    print("\n--- 7. Visualisation des Caractéristiques ---")

    # A. Distribution des features numériques
    print("\nDistribution des features numériques :")
    df[numerical_features].hist(bins=20, figsize=(15, 10))
    plt.suptitle("Distribution des Caractéristiques Numériques")
    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()
    
    # B. Distribution des features catégorielles (par rapport à la cible)
    print("\nDistribution des features catégorielles vs Cible :")
    
    # Ajuster la taille de la grille si nécessaire
    num_cols = 3
    num_rows = (len(categorical_features) + num_cols - 1) // num_cols
    
    fig, axes = plt.subplots(nrows=num_rows, ncols=num_cols, figsize=(16, num_rows * 4))
    axes = axes.flatten() # Pour itérer facilement

    for i, col in enumerate(categorical_features):
        sns.countplot(x=col, data=df, ax=axes[i], hue=TARGET_COL, palette="viridis")
        axes[i].set_title(f"Distribution de '{col}' vs Cible")

    # Cacher les axes superflus
    for j in range(i + 1, len(axes)):
        fig.delaxes(axes[j])
        
    plt.tight_layout()
    plt.show()
```
